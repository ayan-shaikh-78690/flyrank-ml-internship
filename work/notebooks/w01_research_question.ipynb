## 2. The question: decision, action, cost of a wrong call

### Research

**Research Question**

Which webpages should be prioritized for content refresh based on search visibility, engagement, and content freshness?

**Decision**

The goal is to help the content or SEO team decide which webpages should be reviewed and updated first.

**Who takes the action?**

The content editors and SEO team will use the ranked recommendations to decide which pages should be refreshed before others.

**Cost of a wrong recommendation**

If the wrong pages are selected, the team may spend valuable time improving pages with little business impact while important pages continue losing traffic and visibility.

**Why Machine Learning?**

The decision depends on many interacting signals such as impressions, clicks, sessions, CTR, engagement, and content age. These relationships are difficult to capture with simple rules, making machine learning a suitable approach for prioritizing content refresh opportunities.

**Task Type**

Ranking / Scoring

**Target**

## 3. Quick look at the data (2-3 real numbers)

### Quick Look at the Data

The starter dataset contains **30,000 content items** with **44 features** describing search performance, engagement, content characteristics, and freshness. The average number of impressions over the last 90 days is **5,200.37**, while the average number of sessions is **37.07**. Most content items are currently showing a **downward trend (16,262 pages)**, followed by **stable (5,962)** and **upward (4,388)** trends. These statistics suggest that the dataset contains enough variation to identify webpages that should be prioritized for content refresh, making it suitable for the selected research lane.

```python
import pandas as pd

# Load dataset
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Dataset Shape:", df.shape)
print("Total Pages:", len(df))
print("Total Columns:", len(df.columns))

print("\nAverage Impressions:", round(df["impressions_90d"].mean(),2))
print("Average Sessions:", round(df["sessions_90d"].mean(),2))

print("\nTrend Direction Distribution")
print(df["trend_direction"].value_counts())

print("\nTrend Direction (%)")
print((df["trend_direction"].value_counts(normalize=True) * 100).round(2))
```
